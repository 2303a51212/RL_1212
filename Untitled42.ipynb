```python
import random

machines = ['A', 'B', 'C']

N_play = 10

epsilon = 0.2


N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

print("Initial Tables")
print("N =", N)
print("R =", R)
print("Q =", Q)


for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:

        selected_machine = random.choice(machines)
        decision = "Exploration"
    else:

        max_q = max(Q.values())


        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)

    # Part C - Reward Collection
    reward = float(input("Enter reward: "))

    # Part D - Update N
    N[selected_machine] += 1

    # Update total reward R
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", Q)


# Part E - Display Final Results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tQ-value")

for machine in machines:
    print(
        machine,
        "\t", N[machine],
        "\t", R[machine],
        "\t", round(Q[machine], 2)
    )

print("\nFinal Participation Counts:")
print(N)

print("\nFinal Rewards:")
print(R)

print("\nFinal Q-values:")
print(Q)
```
